```python
import timm
import torch

all_gcvit = timm.list_models("*gcvit*")
print(all_gcvit)

all_convnextv2 = timm.list_models("*convnextv2*")
print(all_convnextv2)

gcvit_model = timm.create_model("gcvit_base", pretrained=True)
convnextv2_model = timm.create_model("convnextv2_base", pretrained=True)
print(gcvit_model.num_features)
print(convnextv2_model.num_features)
dummy_img = torch.randn(1, 3, 224, 224)

gcvit_features = gcvit_model(dummy_img)
convnextv2_features = convnextv2_model(dummy_img)
print(gcvit_features.shape)
print(convnextv2_features.shape)

# resolution
print(gcvit_model.default_cfg["input_size"])
print(convnextv2_model.default_cfg["input_size"])
# Output: (3, 224, 224)


# Check all available pre-trained weights for GCViT Base
gcvit_weights = timm.list_models("gcvit_base*", pretrained=True)
print("GCViT Base Weights:")
for w in gcvit_weights:
    print(f" - {w}")

print("\n-------------------\n")

# Check all available pre-trained weights for ConvNeXtV2 Base
convnext_weights = timm.list_models("convnextv2_base*", pretrained=True)
print("ConvNeXtV2 Base Weights:")
for w in convnext_weights:
    print(f" - {w}")
```
